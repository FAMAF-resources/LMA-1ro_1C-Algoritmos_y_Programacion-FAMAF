**Ejercicio 1.** Escribir una función que recibe una cadena de caracteres y devuelve el número de letras `a` que hay en la cadena recibida.

*Ejemplo.* 


```
contar_a('sdaeraad') == 3 # devuelve 3
contar_a('sdf ghhj aaa uiio abab') == 5 # devuelve 5
```





````python
def contar_a_v1(cadena: str) -> int:
  assert type(cadena) == str, "Se debe ingresar una cadena de caracteres"
  return cadena.count('a')

def contar_a(cadena: str) -> int:
  assert type(cadena) == str, "Se debe ingresar una cadena de caracteres"
  res = 0
  # print(list(range(len(cadena))))
  for i in range(len(cadena)): 
    if cadena[i] == 'a':
      res = res +  1
  return  res

print(contar_a('sdaeraad'))  # devuelve 3
print(contar_a('sdf ghhj aaa uiio abab')) # devuelve 5

def contar_char(car, cadena: str) -> int:
  # cantidad de veces que el caracter car está en la cadena
  # car = str(car)
  assert type(car) == type(cadena) == str and len(car) == 1, 'Deben ser cadenas y len(car) == 1'
  res = 0
  for i in range(len(cadena)): 
    if cadena[i] == car:
      res = res +  1
  return  res

print(contar_char('a', 'sdaeraad'))  # devuelve 3
print(contar_char('a', 'sdf ghhj aaa uiio abab')) # devuelve 5
print(contar_char('d', 'sdaeraad'))  # devuelve 3
print(contar_char('d', 'sdf ghhj aaa uiio abab')) # devuelve 5
print(contar_char('3', 'sd3a3er3aad'))  # devuelve 3
# print(contar_char(3, 'sdf g3hhj a3aa uiio abab')) # devuelve 5
````

**Ejercicio 3.** Escribir una función que recibe una cadena de caracteres y devuelve una lista de 26 números. El primero de esos números (el que se encuentra en la posición 0 de la lista) es el número de letras `a`. El segundo es el número de letras `b`, etc. El último, es el número de letras `z`.

*Observación.* La posición 0 de la lista que se devuelve corresponde a la letra `a`, la posición 1 corresponde a la letra `b`, etc.

En general las posiciones de las letras está dadas por 
```
letras = ['a', 'b', 'c', 'd', 'e', 'f', 'g', 'h', 'i', 'j', 'k', 'l', 'm', 'n', 'o', 'p', 'q', 'r', 's', 't', 'u', 'v', 'w', 'x', 'y', 'z']
```

*Ejemplo.* 
```
contar_letras('sdaeraad') = [3, 0, 0, 2, 1, 0, ...] 
```



````python
letras = ['a', 'b', 'c', 'd', 'e', 'f', 'g', 'h', 'i', 'j', 'k', 'l', 'm', 'n', 'o', 'p', 'q', 'r', 's', 't', 'u', 'v', 'w', 'x', 'y', 'z']

-
def contar_letras(cadena: str) -> list:
  res = []
  for letra in letras:
    num = cadena.count(letra)
    res.append(num)
  return res

print(contar_letras('sdaeraad'))


# Esta función recorre 26 veces la cadena, una vez por cada letra (pues count() 
# en su implementación recorre la cadena)
````

````python
# Usando  la función ord(), que nos da la posición del caracter (según Python), 
# se puede hacer una función que recorra una sola vez la cadena.

def contar_letras_v2(cadena: str) -> list:
  res = [0]*26 # lista de 26 ceros
  for letra in cadena:
    ind = ord(letra) - 97 # ord('a') = 97, ord('b') = 98, etc.
    res[ind] = res[ind] + 1 
  return res

print(contar_letras_v2('sdaeraad'))
````
