**Ejercicio 1.** $x^n$: elevar un número $x$ a una potencia entera no negativa se puede resolver con un ciclo for que devuelva $x \cdot x \cdots\ x\cdot  x$, $n$ veces, pero en este ejercicio se pide resolverlo definiendo una función que utiliza "recursión en $n$", considerando $2$ casos: cuando $n$ es $0$ y cuando $n$ es mayor que $0$. Luego de resolverlo, resulta interesante pensar en la precondición de esta función: ¿qué debería satisfacer $n$? ¿qué debería satisfacer $x$? ¿qué ocurre si ambos son $0$?

```python
# 0**0 debería ser 1

def potencia_iter(x, n):
  # pre: x es int o float, n es int  <= 0
  # post: devuelve x**n
  assert (type(x) == int or type(x) == float) and type(n) == int and n >= 0, 'Error' 
  res = 1
  for _ in range(n):
    res = res * x
  return res

print(potencia_iter(2.5,19))
print(2.5**19)

# potencia_iter() no es lo que pide el ejercicio

def potencia(x, n):
  # pre: x es int o float, n es int  <= 0
  # post: devuelve x**n
  assert (type(x) == int or type(x) == float) and type(n) == int and n >= 0, 'Error' 
  if n == 0:
    res = 1
  else:
    res = potencia(x, n - 1) * x
  return res

print(potencia(3,4))
```


**Ejercicio 2.** Dados $n$ y $m$  enteros no negativos, dar una función recursiva que divida $n$ por $m$ y devuelva su cociente (sin usar `//`, ni `/`). Los dos casos a considerar son $n < m$ y $n \ge m$. Nuevamente es interesante pensar en la precondición. Dar una función recursiva que devuelva el resto de la división (sin usar `%`).

```python
# Tener presente que 5 * n = n + n + n + n + n, es decir sumar n 5 veces. 
# La división es una "resta generalizada"
# Ejemplo: 23 // 5 = 4, 
# 1. 23 - 5 = 18 (23//5=4)
# 2. 18 - 5 = 13 (18//5=3)
# 3. 13 - 5 = 8  (13//5=2)
# 4.  8 - 5 = 3
# Luego 23 = 5 * 4 + 3
# se decir n // m = q si restando q veces m obtengo un r tal que 0 <= r < m

# Idea: division(n, m) = division(n - m, m) + 1  

def division(n, m):
  # pre: n,m enteros n >=0, m > 0
  # post devuelve n // m
  assert type(n) == type(m) == int and n >= 0 and m > 0, 'Error'
  if 0 <= n < m:
    res = 0
  else:
    res = division(n - m, m) + 1
  return res  

division(23,5)
division(18,5)

# Extendamos la división a n entero cualquiera

def division(n, m):
  # pre: n,m enteros m > 0
  # post devuelve n // m
  assert type(n) == type(m) == int and m > 0, 'Error'

  if 0 <= n < m:
    res = 0
  elif n < 0:
    res = division(n + m, m) - 1
  else:
    res = division(n - m, m) + 1
  
  return res  

# Una solución "fea":
def division2(n, m):
  # pre: n,m enteros m > 0
  # post devuelve n // m
  assert type(n) == type(m) == int and m > 0, 'Error'

  if n < 0:
    np = -n
  if np < m:
    res = 0
  else:
    res = division(np - m, m) + 1
  
  if n < 0:
    res = - res -1  
  return res 

division(-5,2)
division2(-5,2)
```

```python
# Recordemos que n = (n // m) * m + (n % m) => n % m = n - (n // m) * m 
# En nuestro caso resto(n, m) = n - division(n, m) * m

def resto(n, m):
  # pre: n,m enteros m > 0
  # post devuelve n // m
  assert type(n) == type(m) == int and m > 0, 'Error'
  resturn n - division(n, m) * m 
```

**Ejercicio 3.** Ahora una definición que devuelva ambas cosas: el cociente y el resto. Para ello podés utilizar `return a, b` para devolver dos valores. En la llamada recursiva podés escribir `x, y = div(n, m)` para que guarde el primer valor que devuelve en `x`, el segundo en `y`.

```python
# div(n, m) = (n // m, n % m) = ((n - m) // m + 1, (n - m) % m)

# 5 = 7 * 0 + 5
# 7 = 5 * 1 + 2
def div(n, m):
  # pre: n,m enteros n >=0, m > 0
  # post devuelve (n // m, n % m)
  assert type(n) == type(m) == int and n >= 0 and m > 0, 'Error'

  if n < m:
    res = (0, n)
  else: 
    div_r = div(n - m, m) # llamada recursiva
    res = (div_r[0] + 1, div_r[1]) # ¿Se puede poner div(n - m, m) + (1,0)? 
    # No (el + de tuplas es concatenación)

  return res

div(5,7)
div(7,5)
div(23,5)
```

**Ejercicio 6.** Para $n$ entero no negativo y $m$ entero entre $0$ y $n$ se define el número combinatorio $\displaystyle\binom{n}{m}$ o coeficiente binomial, que es de gran importancia en matemática. 

Aparece en el desarrollo de la potencia del binomio: 
$$
(a+b)^n = \sum_{k=0}^n  \binom{n}{k} a^{n-k}b^{k},
$$
 de donde surge el nombre *coeficiente binomial*. También se utiliza extensivamente en el cálculo de probabilidades puesto que  $\displaystyle\binom{n}{m}$ mide el número de combinaciones (de allí *número combinatorio*) posibles de  $m$ elementos eligiendo libremente de un conjunto donde hay $n$ elementos.
 
Un definición que podemos encontrar en la literatura del número combinatorio es la siguiente:
$$
\binom{n}{m} = \frac{n!}{(n-m)!m!}.
$$

El teorema de Pascal dice que ese número también se puede calcular de manera recursiva. Por la fórmula anterior $\displaystyle\binom{n}{0} = \binom{n}{n} = 1$. Si $m>0$, entonces $n$ también debe serlo pues $n \ge m$. En tal caso el teorema dice que
$$
\binom{n}{m} = \binom{n-1}{m-1} + \binom{n-1}{m}. 
$$
Utilizar este teorema para dar una definición recursiva del número combinatorio . ¿Cuál es la precondición?

**Observación.** Calculesmos $b^0$ con $b \ne 0$ de dos formas. La primera es por definicón $b^0 = 1$. 

La segunda es por binomio de Newton:
$$
(0+b)^0 = \sum_{k=0}^0  \binom{0}{k} 0^{0-k}b^{k},
$$
es decir
$$
(0+b)^0 = \binom{0}{0} 0^{0-0}b^{0} = \binom{0}{0} 0^{0}b^{0} = 0^{0}b^{0}.
$$
Para se coherentes,  el valor de $0^0$  debe ser 1. 


```python
def binom(n, m):
  # pre: n >= m >=0 
  # post:  devuelve (n m) el número combinatorio
  if n == m or m == 0:
    res = 1
  else:
    res = binom(n - 1, m - 1) + binom(n - 1, m)
  return res

binom(12,7)

```
